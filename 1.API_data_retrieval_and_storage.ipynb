##### Fetch Data from Open Library REST API

```python
import requests
import sqlite3

url="https://openlibrary.org/subjects/fantasy.json?limit=10" 

response=requests.get(url)

if response.status_code==200:
    data=response.json()
    books=data["works"]
else:
    print("Failed")
    exit()
```

```python
print(data)

print("\n",books[2:])
```

##### Connect to SQLite Database

```python
conn=sqlite3.connect("Books.db")

cursor=conn.cursor()
```

```python
#create books table

cursor.execute(""" CREATE TABLE IF NOT EXISTS books (
        id integer PRIMARY KEY AUTOINCREMENT,
        title text,
        author text,
        publication_year integer
    )""")
```

##### Insert API data into database

```python
for book in books:

    title=book.get("title","Unknown")

    authors=book.get("authors",[])
    
    author=authors[0].get("name", "Unknown") if authors else "Unknown"

    publication_year=book.get("first_publish_year",None)

    cursor.execute("""
        INSERT INTO books (title, author, publication_year)
        VALUES (?,?,?)
    """, (title,author,publication_year))
```

```python
conn.commit()
```

```python
cursor.execute("SELECT * FROM books")

rows=cursor.fetchall()

for row in rows:
    print(f"ID: {row[0]}")
    print(f"Title: {row[1]}")
    print(f"Author: {row[2]}")
    print(f"Publication Year: {row[3]}")
    print("\n\n")
```

```python
conn.close()
```
